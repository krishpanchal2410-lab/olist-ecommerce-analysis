```python
from pathlib import Path

possible_paths = [
    Path.cwd() / "data" / "raw",
    Path.cwd().parent / "data" / "raw",
]

raw_dir = next((path for path in possible_paths if path.exists()), None)

if raw_dir is None:
    print("Can't find data/raw yet.")
else:
    csv_files = sorted(raw_dir.glob("*.csv"))
    print(f"Found {len(csv_files)} CSV files:")
    for file in csv_files:
        print(file.name)
```
